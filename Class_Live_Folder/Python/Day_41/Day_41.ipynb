```python
data = {"names":["AAA","PFS","JFS","DA","DS"],"ids":[12,43,56,78,67]}
f = pd.Series(data)
print(f)
print(f.index)
print(f.dtype)
print(f['names'])
print(f['ids'])
print(f['names'][2])
print(f['ids'][:2])
print(f['ids'])
```
